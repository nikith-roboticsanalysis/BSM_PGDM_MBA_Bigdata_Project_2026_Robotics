```python

import pandas as pd
from pathlib import Path

# Load the robotics dataset uploaded to Colab
file_path = "/content/global_robotics_market_intelligence_master.csv"

df = pd.read_csv(file_path)

# Display the first five records
display(df.head())
```

```python

print("Number of rows and columns:", df.shape)

print("\nColumn names:")
print(df.columns.tolist())

print("\nMissing values:")
display(df.isnull().sum())

print("\nDuplicate rows:", df.duplicated().sum())

print("\nData types:")
display(df.dtypes)
```

```python

# Number of rows and columns
print(f"Total records: {df.shape[0]:,}")
print(f"Total columns: {df.shape[1]}")

# Column names
print("\nColumn names:")
for column in df.columns:
    print("-", column)
```

```python

# Count missing values in each column
missing_values = df.isnull().sum()

print("Missing values per column:")
display(missing_values[missing_values > 0])

if missing_values.sum() == 0:
    print("No missing values found.")

# Count duplicate records
duplicate_count = df.duplicated().sum()
print(f"\nDuplicate rows: {duplicate_count:,}")
```

```python

# Summary statistics for numeric columns
display(df.describe().T)
```

```python

# Display the data type and number of unique values
column_summary = pd.DataFrame({
    "Data Type": df.dtypes.astype(str),
    "Unique Values": df.nunique(),
    "Missing Values": df.isnull().sum()
})

display(column_summary)
```

```python

from pathlib import Path

output_dir = Path("/content/outputs")
output_dir.mkdir(exist_ok=True)

# Save a summary of the dataset structure
column_summary.to_csv(
    output_dir / "dataset_column_summary.csv",
    index=True
)

print("Column summary saved to:", output_dir / "dataset_column_summary.csv")
```

```python

# Step 10.1: Revenue and profitability KPIs

revenue_col = "revenue_usd"
cost_col = "estimated_cost_usd"
profit_col = "estimated_gross_profit_usd"

total_revenue = df[revenue_col].sum()
total_cost = df[cost_col].sum()
total_profit = df[profit_col].sum()

profit_margin = (
    total_profit / total_revenue * 100
    if total_revenue != 0 else 0
)

kpi_summary = pd.DataFrame({
    "Metric": [
        "Total recorded revenue (USD-labelled)",
        "Total estimated cost (USD-labelled)",
        "Total estimated gross profit (USD-labelled)",
        "Estimated gross profit margin (%)",
        "Number of records"
    ],
    "Value": [
        total_revenue,
        total_cost,
        total_profit,
        profit_margin,
        len(df)
    ]
})

display(kpi_summary.round(2))
```

```python

category_analysis = (
    df.groupby("product_category", dropna=False)
      .agg(
          records=("transaction_id", "count"),
          total_revenue=("revenue_usd", "sum"),
          estimated_cost=("estimated_cost_usd", "sum"),
          estimated_profit=("estimated_gross_profit_usd", "sum"),
          total_quantity=("quantity", "sum")
      )
      .reset_index()
)

category_analysis["profit_margin_pct"] = (
    category_analysis["estimated_profit"]
    / category_analysis["total_revenue"].replace(0, float("nan"))
    * 100
)

category_analysis = category_analysis.sort_values(
    "total_revenue", ascending=False
)

display(category_analysis.round(2))
```

```python

import matplotlib.pyplot as plt

top_categories = category_analysis.head(10).sort_values(
    "total_revenue", ascending=True
)

plt.figure(figsize=(10, 6))
plt.barh(
    top_categories["product_category"].astype(str),
    top_categories["total_revenue"]
)
plt.title("Top 10 Product Categories by Recorded Revenue")
plt.xlabel("Recorded Revenue (USD-labelled)")
plt.ylabel("Product Category")
plt.tight_layout()
plt.show()
```

```python

# Parse dates using day-first format
df["order_date"] = pd.to_datetime(
    df["order_date"], dayfirst=True, errors="coerce"
)

monthly_analysis = (
    df.dropna(subset=["order_date"])
      .assign(month=lambda x: x["order_date"].dt.to_period("M").astype(str))
      .groupby("month")
      .agg(
          revenue=("revenue_usd", "sum"),
          estimated_profit=("estimated_gross_profit_usd", "sum")
      )
      .reset_index()
)

plt.figure(figsize=(12, 5))
plt.plot(
    monthly_analysis["month"],
    monthly_analysis["revenue"],
    marker="o",
    label="Recorded revenue"
)
plt.title("Monthly Recorded Revenue Trend")
plt.xlabel("Month")
plt.ylabel("Recorded Revenue (USD-labelled)")
plt.xticks(rotation=60)
plt.legend()
plt.tight_layout()
plt.show()

plt.figure(figsize=(12, 5))
plt.plot(
    monthly_analysis["month"],
    monthly_analysis["estimated_profit"],
    marker="o",
    label="Estimated gross profit"
)
plt.title("Monthly Estimated Gross Profit Trend")
plt.xlabel("Month")
plt.ylabel("Estimated Gross Profit (USD-labelled)")
plt.xticks(rotation=60)
plt.legend()
plt.tight_layout()
plt.show()
```

```python

from pathlib import Path

output_dir = Path("/content/outputs")
output_dir.mkdir(exist_ok=True)

kpi_summary.to_csv(output_dir / "financial_kpis.csv", index=False)
category_analysis.to_csv(
    output_dir / "category_profitability.csv", index=False
)
monthly_analysis.to_csv(
    output_dir / "monthly_revenue_profit.csv", index=False
)

print("Analysis tables saved in:", output_dir)
```

```python

# Step 11.1: Prepare analysis columns

required_columns = [
    "country_name",
    "industry",
    "sales_channel",
    "revenue_usd",
    "estimated_cost_usd",
    "estimated_gross_profit_usd"
]

missing_columns = [
    col for col in required_columns if col not in df.columns
]

if missing_columns:
    raise ValueError(f"Missing columns: {missing_columns}")

# Convert financial columns to numeric values
financial_columns = [
    "revenue_usd",
    "estimated_cost_usd",
    "estimated_gross_profit_usd"
]

for col in financial_columns:
    df[col] = pd.to_numeric(df[col], errors="coerce")

print("Dataset ready for analysis.")
print("Records:", f"{len(df):,}")
```

```python

country_analysis = (
    df.groupby("country_name", dropna=False)
      .agg(
          records=("revenue_usd", "count"),
          total_revenue=("revenue_usd", "sum"),
          estimated_cost=("estimated_cost_usd", "sum"),
          estimated_profit=("estimated_gross_profit_usd", "sum")
      )
      .reset_index()
)

country_analysis["profit_margin_pct"] = (
    country_analysis["estimated_profit"]
    / country_analysis["total_revenue"].replace(0, float("nan"))
    * 100
)

country_analysis = country_analysis.sort_values(
    "total_revenue", ascending=False
)

display(country_analysis.round(2))
```

```python

import matplotlib.pyplot as plt

top_countries = country_analysis.head(10).sort_values(
    "total_revenue", ascending=True
)

plt.figure(figsize=(10, 6))
plt.barh(
    top_countries["country_name"].fillna("Unknown").astype(str),
    top_countries["total_revenue"]
)
plt.title("Top 10 Countries by Recorded Revenue")
plt.xlabel("Recorded Revenue (USD-labelled)")
plt.ylabel("Country")
plt.tight_layout()
plt.show()
```

```python

industry_analysis = (
    df.groupby("industry", dropna=False)
      .agg(
          records=("revenue_usd", "count"),
          total_revenue=("revenue_usd", "sum"),
          estimated_cost=("estimated_cost_usd", "sum"),
          estimated_profit=("estimated_gross_profit_usd", "sum"),
          total_quantity=("quantity", "sum")
      )
      .reset_index()
)

industry_analysis["profit_margin_pct"] = (
    industry_analysis["estimated_profit"]
    / industry_analysis["total_revenue"].replace(0, float("nan"))
    * 100
)

industry_analysis = industry_analysis.sort_values(
    "total_revenue", ascending=False
)

display(industry_analysis.round(2))
```

```python

industry_plot = industry_analysis.sort_values(
    "total_revenue", ascending=True
)

plt.figure(figsize=(10, 6))
plt.barh(
    industry_plot["industry"].fillna("Unknown").astype(str),
    industry_plot["total_revenue"]
)
plt.title("Recorded Revenue by Industry")
plt.xlabel("Recorded Revenue (USD-labelled)")
plt.ylabel("Industry")
plt.tight_layout()
plt.show()
```

```python

channel_analysis = (
    df.groupby("sales_channel", dropna=False)
      .agg(
          records=("revenue_usd", "count"),
          total_revenue=("revenue_usd", "sum"),
          estimated_cost=("estimated_cost_usd", "sum"),
          estimated_profit=("estimated_gross_profit_usd", "sum"),
          total_quantity=("quantity", "sum")
      )
      .reset_index()
)

channel_analysis["profit_margin_pct"] = (
    channel_analysis["estimated_profit"]
    / channel_analysis["total_revenue"].replace(0, float("nan"))
    * 100
)

channel_analysis = channel_analysis.sort_values(
    "total_revenue", ascending=False
)

display(channel_analysis.round(2))
```

```python

channel_plot = channel_analysis.sort_values(
    "total_revenue", ascending=True
)

plt.figure(figsize=(9, 5))
plt.barh(
    channel_plot["sales_channel"].fillna("Unknown").astype(str),
    channel_plot["total_revenue"]
)
plt.title("Recorded Revenue by Sales Channel")
plt.xlabel("Recorded Revenue (USD-labelled)")
plt.ylabel("Sales Channel")
plt.tight_layout()
plt.show()
```

```python

import pandas as pd
from pathlib import Path

# 1. Locate the analysis results
output_dir = Path("/content/outputs")
output_dir.mkdir(exist_ok=True)

result_files = {
    "KPIs": "financial_kpis.csv",
    "Categories": "category_profitability.csv",
    "Countries": "country_analysis.csv",
    "Industries": "industry_analysis.csv",
    "Sales Channels": "sales_channel_analysis.csv",
    "Monthly Trends": "monthly_revenue_profit.csv"
}

results = {}

for name, filename in result_files.items():
    file_path = output_dir / filename
    if file_path.exists():
        results[name] = pd.read_csv(file_path)
        print(f"Loaded: {filename}")
    else:
        print(f"Not found yet: {filename}")

# 2. Display the overall financial KPIs
if "KPIs" in results:
    print("\n--- OVERALL FINANCIAL KPIs ---")
    display(results["KPIs"])

# 3. Find the leading group in each analysis
def show_top_group(name, dataframe):
    print(f"\n--- {name.upper()} ---")
    display(dataframe.head(5))

for name in ["Categories", "Countries", "Industries", "Sales Channels"]:
    if name in results:
        show_top_group(name, results[name])

# 4. Create a combined analytical summary
report = []
report.append("GLOBAL ROBOTICS MARKET INTELLIGENCE")
report.append("COMBINED ANALYTICAL SUMMARY")
report.append("=" * 45)
report.append("")
report.append("Dataset scope: 50,000 records, subject to notebook validation.")
report.append(
    "Important limitation: This dataset is synthetic/illustrative. "
    "Its results should not be presented as verified real-world market figures."
)
report.append(
    "Currency limitation: Validate currency_code before comparing financial "
    "totals across countries, because the dataset contains multiple currencies."
)
report.append("")

if "KPIs" in results:
    report.append("1. OVERALL FINANCIAL KPIs")
    report.append(results["KPIs"].to_string(index=False))
    report.append("")

sections = [
    ("Categories", "2. PRODUCT CATEGORY ANALYSIS"),
    ("Countries", "3. COUNTRY ANALYSIS"),
    ("Industries", "4. INDUSTRY ANALYSIS"),
    ("Sales Channels", "5. SALES-CHANNEL ANALYSIS"),
    ("Monthly Trends", "6. MONTHLY REVENUE AND PROFIT TRENDS")
]

for key, heading in sections:
    if key in results:
        report.append(heading)
        report.append(results[key].head(10).to_string(index=False))
        report.append("")

report.append("INTERPRETATION GUIDANCE")
report.append(
    "- Compare revenue, estimated gross profit, and profit margin together."
)
report.append(
    "- High revenue does not necessarily mean high profitability."
)
report.append(
    "- Treat country comparisons cautiously until currencies are standardized."
)
report.append(
    "- Findings describe this dataset and do not establish actual market share "
    "or real-world robotics industry performance."
)

summary_path = output_dir / "combined_analytical_summary.txt"
summary_path.write_text("\n".join(report), encoding="utf-8")

print("\nCombined summary saved to:", summary_path)
print("\n--- SUMMARY PREVIEW ---")
print(summary_path.read_text(encoding="utf-8")[:5000])
```
